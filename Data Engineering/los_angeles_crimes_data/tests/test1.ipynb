# Test 1

This test is to check if the FY 2025 Disability Compensation Recipients by County dataset which i found in the internet, is good enough for usage. 

```python
import requests
import pandas as pd

# API endpoint
url = "https://www.data.va.gov/api/v3/views/95an-zhhy/query.json"

params = {
    "accessType": "DOWNLOAD"
}

headers = {
    "Accept": "application/json"
}

# Send GET request
response = requests.get(
    url,
    headers=headers,
    params=params,
    timeout=30
)

# Check if request succeeded
response.raise_for_status()

# Convert JSON response to Python data
data = response.json()

# Convert to pandas DataFrame
df = pd.DataFrame(data)

print(df.head())
print(df.shape)
print(df.columns)
```
