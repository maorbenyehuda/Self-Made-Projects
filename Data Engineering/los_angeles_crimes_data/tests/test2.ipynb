This test is to check if the Crime Data from 2020 to 2024 in LA dataset which i found in the internet, is good enough for usage. 

```python
import requests
import pandas as pd

# API endpoint
url = "https://data.lacity.org/api/v3/views/2nrs-mtv8/query.json"

params = {
    "accessType": "DOWNLOAD"
}



# Send GET request
response = requests.get(
    url,
    params=params,
    timeout=30
)

# Check if request succeeded
response.raise_for_status()

# Convert JSON response to Python data
data = response.json()

# Convert to pandas DataFrame
df = pd.DataFrame(data)

print(df.head())
print(df.shape)
print(df.columns)
```
